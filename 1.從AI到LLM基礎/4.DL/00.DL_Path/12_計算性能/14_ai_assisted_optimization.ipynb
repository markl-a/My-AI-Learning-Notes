## 1. AI 優化提示詞模板

### 模板 1: Profiling 結果分析

````python
def generate_profiling_analysis_prompt(prof_data):
    """生成 Profiling 分析提示詞"""
    prompt = f"""
我正在優化一個深度學習模型的性能。以下是 PyTorch Profiler 的結果：

```
{prof_data}
```

請幫我分析：

1. **主要瓶頸識別**：
   - 哪個操作佔用時間最多？
   - CPU 還是 GPU 是瓶頸？
   - 是否存在記憶體傳輸瓶頸？

2. **優化建議（優先級排序）**：
   - 列出 3-5 個最有效的優化方向
   - 每個建議估計可能的性能提升
   - 提供具體的實現代碼片段

3. **潛在風險**：
   - 哪些優化可能影響模型準確度？
   - 哪些優化需要大量工程工作？

請提供實用、可立即實施的建議。
    """
    return prompt

print("✓ Profiling 分析提示詞模板已定義")
````

### 模板 2: 代碼優化建議

````python
def generate_code_optimization_prompt(code_snippet, issue):
    """生成代碼優化提示詞"""
    prompt = f"""
我的深度學習訓練代碼存在以下問題：{issue}

當前代碼：
```python
{code_snippet}
```

請幫我：
1. 識別代碼中的性能問題
2. 提供優化後的代碼
3. 解釋優化的原理
4. 說明預期的性能提升

要求：
- 保持代碼的功能不變
- 優先考慮簡單且有效的優化
- 如果需要權衡，請明確說明
    """
    return prompt

print("✓ 代碼優化提示詞模板已定義")
````

### 模板 3: 硬體配置建議

````python
def generate_hardware_recommendation_prompt(model_info, budget):
    """生成硬體配置建議提示詞"""
    prompt = f"""
我需要為以下深度學習任務選擇硬體：

模型信息：
{model_info}

預算：{budget}

請建議：
1. **GPU 選擇**：
   - 推薦的 GPU 型號和數量
   - 性價比分析
   
2. **其他配置**：
   - CPU、RAM、存儲建議
   - 網絡配置（如果是分佈式訓練）

3. **雲端 vs 本地**：
   - 成本對比
   - 適用場景分析

4. **優化策略**：
   - 在此硬體上的最佳實踐
   - 預期訓練時間
    """
    return prompt

print("✓ 硬體配置建議提示詞模板已定義")
````

## 3. 自動生成性能報告

````python
def generate_performance_report(before_metrics, after_metrics):
    """生成性能優化報告"""
    report = f"""
# 性能優化報告

## 優化前
- 訓練速度: {before_metrics['speed']} images/sec
- GPU 利用率: {before_metrics['gpu_util']}%
- 記憶體使用: {before_metrics['memory']} GB

## 優化後  
- 訓練速度: {after_metrics['speed']} images/sec
- GPU 利用率: {after_metrics['gpu_util']}%
- 記憶體使用: {after_metrics['memory']} GB

## 改進
- 速度提升: {after_metrics['speed'] / before_metrics['speed']:.2f}x
- GPU 利用率提升: {after_metrics['gpu_util'] - before_metrics['gpu_util']}%
- 記憶體節省: {(1 - after_metrics['memory'] / before_metrics['memory']) * 100:.1f}%

## 實施的優化
1. 啟用混合精度訓練 (AMP)
2. 優化數據加載流程
3. 調整批次大小

## 後續建議
- 考慮使用 torch.compile() 進一步加速
- 探索分佈式訓練（如需要更大規模）
- 定期 profiling 監控性能
    """
    return report

# 示例使用
before = {'speed': 100, 'gpu_util': 30, 'memory': 8.0}
after = {'speed': 280, 'gpu_util': 85, 'memory': 5.5}

report = generate_performance_report(before, after)
print(report)
````
