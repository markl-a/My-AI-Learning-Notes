# 循環神經網路
:label:`chap_rnn`

到目前為止，我們遇到過兩種類型的數據：表格數據和圖像數據。
對於圖像數據，我們設計了專門的卷積神經網路架構來為這類特殊數據結構建模。
換句話說，如果我們擁有一張圖像，我們需要有效地利用其像素位置，
假若我們對圖像中的像素位置進行重排，就會對圖像中內容的推斷造成極大的困難。

最重要的是，到目前為止我們預設數據都來自某種分布，
並且所有樣本都是獨立同分布的
（independently and identically distributed，i.i.d.）。
然而，大多數的數據並非如此。
例如，文章中的單字是按順序寫的，如果順序被隨機地重排，就難以理解文章原始的意思。
同樣，視頻中的圖像幀、對話中的音頻信號以及網站上的瀏覽行為都是有順序的。
因此，針對此類數據而設計特定模型，可能效果會更好。

另一個問題來自這樣一個事實：
我們不僅可以接收一個序列作為輸入，而是還可能期望繼續猜測這個序列的後續。
例如，一個任務可以是繼續預測$2, 4, 6, 8, 10, \ldots$。
這在時間序列分析中是相當常見的，可以用來預測股市的波動、
患者的體溫曲線或者賽車所需的加速度。
同理，我們需要能夠處理這些數據的特定模型。

簡言之，如果卷積神經網路可以有效地處理空間信息，
那麼本章的*循環神經網路*（recurrent neural network，RNN）則可以更好地處理序列信息。
循環神經網路通過引入狀態變量來存儲過去的資訊和當前的輸入，從而可以確定當前的輸出。

許多使用循環網路的例子都是基於文本數據的，因此我們將在本章中重點介紹語言模型。
在對序列數據進行更詳細的回顧之後，我們將介紹文本預處理的實用技術。
然後，我們將討論語言模型的基本概念，並將此討論作為循環神經網路設計的靈感。
最後，我們描述了循環神經網路的梯度計算方法，以探討訓練此類網路時可能遇到的問題。
