# 現代卷積神經網路
:label:`chap_modern_cnn`

上一章我們介紹了卷積神經網路的基本原理，本章將介紹現代的卷積神經網路架構，許多現代卷積神經網路的研究都是建立在這一章的基礎上的。
在本章中的每一個模型都曾一度佔據主導地位，其中許多模型都是ImageNet競賽的優勝者。ImageNet競賽自2010年以來，一直是計算機視覺中監督學習進展的指向標。

## 📚 完整學習路徑

本章分為兩大部分：**經典架構基礎**（2012-2017）和**實戰與進階技術**（2017-2024）。

### 第一部分：經典架構（2012-2017）

這些模型奠定了現代深度學習的基礎：

- **AlexNet** (2012)。它是第一個在大規模視覺競賽中擊敗傳統計算機視覺模型的大型神經網路，開啟了深度學習革命；
- **VGG** (2014)。它利用許多重複的神經網路塊，證明了網路深度的重要性；
- **NiN** (2014)。網路中的網路，它重複使用$1\times 1$卷積層來增強非線性和減少參數；
- **GoogLeNet** (2014)。含並行連結的網路，通過Inception模塊並行抽取多尺度信息；
- **Batch Normalization** (2015)。批量歸一化技術，大幅提升訓練穩定性和速度；
- **ResNet** (2015)。殘差網路通過跨層連接構建超深網路，是最具影響力的架構之一；
- **DenseNet** (2017)。稠密連接網路，通過密集連接實現極致的特徵重用。

### 第二部分：實戰與現代技術（2017-2024）

從理論到實踐，掌握最新技術：

- **遷移學習** (2024新增)。深度學習最實用的技術，學習如何使用預訓練模型、Fine-tuning策略、以及timm庫的使用；
- **現代架構** (2024新增)。探索2016年後的重要創新：
  - SENet：通道注意力機制
  - MobileNet系列：輕量級網路設計
  - EfficientNet：複合縮放與神經架構搜索
- **模型可視化** (2024新增)。理解模型決策過程：GradCAM、特徵可視化、濾波器分析、誤分類診斷。

## 💡 本章亮點

雖然深度神經網路的概念非常簡單——將神經網路堆疊在一起。但由於不同的網路架構和超參數選擇，這些神經網路的性能會發生很大變化。

本章的特色：
- ✅ **完整歷史脈絡**：從2012到2024，見證CNN的演進
- ✅ **理論與實踐結合**：不只是論文，更有完整代碼實現
- ✅ **AI輔助工具**：timm、GradCAM、Captum等現代工具庫
- ✅ **實戰導向**：遷移學習、模型優化、可視化調試
- ✅ **最佳實踐**：工業級的技巧和建議
