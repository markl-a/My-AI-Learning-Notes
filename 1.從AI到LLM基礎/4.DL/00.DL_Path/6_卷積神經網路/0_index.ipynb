# 卷積神經網路
:label:`chap_cnn`

在前面的章節中，我們遇到過圖像數據。
這種數據的每個樣本都由一個二維像素網格組成，
每個像素可能是一個或者多個數值，取決於是黑白還是彩色圖像。
到目前為止，我們處理這類結構豐富的數據的方式還不夠有效。
我們僅僅通過將圖像數據展平成一維向量而忽略了每個圖像的空間結構信息，再將數據送入一個全連接的多層感知機中。
因為這些網路特徵元素的順序是不變的，因此最優的結果是利用先驗知識，即利用相近像素之間的相互關聯性，從圖像數據中學習得到有效的模型。

本章介紹的*卷積神經網路*（convolutional neural network，CNN）是一類強大的、為處理圖像數據而設計的深度學習模型。
基於卷積神經網路架構的模型在計算機視覺領域中已經占主導地位，當今幾乎所有的圖像識別、目標檢測或語義分割相關的學術競賽和商業應用都以這種方法為基礎。

現代卷積神經網路架構的設計得益於生物學、群論和一系列的補充實驗。
卷積神經網路需要的參數少於全連接架構的網路，而且卷積也很容易用GPU並行計算。
因此卷積神經網路除了能夠高效地採樣從而獲得精確的模型，還能夠高效地計算。
久而久之，從業人員越來越多地使用卷積神經網路。即使在通常使用循環神經網路的一維序列結構任務上（例如音頻、文本和時間序列分析），卷積神經網路也越來越受歡迎。
通過對卷積神經網路一些巧妙的調整，也使它們在圖結構數據和推薦系統中發揮作用。

在本章的開始，我們將介紹構成所有卷積網路主幹的基本元素。
這包括卷積層本身、填充（padding）和步幅（stride）的基本細節、用於在相鄰區域彙集信息的彙集層（pooling）、在每一層中多通道（channel）的使用，以及有關現代卷積網路架構的仔細討論。
在本章的最後，我們將介紹一個完整的、可運行的LeNet模型：這是第一個成功應用的卷積神經網路，比現代深度學習興起時間還要早。
在下一章中，我們將深入研究一些流行的、相對較新的卷積神經網路架構的完整實現，這些架構涵蓋了現代從業者通常使用的大多數經典技術。
