# 線性神經網路
:label:`chap_linear` 

在介紹深度神經網路之前，我們需要了解神經網路訓練的基礎知識。
本章我們將介紹神經網路的整個訓練過程，
包括：定義簡單的神經網路架構、資料處理、指定損失函數和如何訓練模型。
為了更容易學習，我們將從經典演算法————*線性*神經網路開始，介紹神經網路的基礎知識。
經典統計學習技術中的線性回歸和softmax回歸可以視為線性神經網路，
這些知識將為本書其他部分中更複雜的技術奠定基礎。
