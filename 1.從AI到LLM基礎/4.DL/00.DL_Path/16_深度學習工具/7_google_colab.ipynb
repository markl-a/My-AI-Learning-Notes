# 使用 Google Colab
:label:`sec_colab`

Google Colab（Colaboratory）是一個免費的雲端 Jupyter notebook 環境，無需任何設置即可使用，完全在雲端運行。它特別適合深度學習初學者和研究人員，因為它提供免費的 GPU 和 TPU 訪問權限。

## Google Colab 簡介

### 主要特點

Google Colab 提供以下優勢：

1. **免費 GPU/TPU 訪問**：無需購買硬體即可進行深度學習訓練
2. **無需安裝**：直接在瀏覽器中運行，無需本地配置
3. **協作功能**：類似 Google Docs，可以輕鬆分享和協作
4. **預安裝環境**：已預裝常用的深度學習框架
5. **整合 Google Drive**：方便數據存儲和管理

### 訪問 Colab

訪問 [https://colab.research.google.com](https://colab.research.google.com) 即可開始使用。你需要一個 Google 帳號來登入。

## 免費 GPU/TPU 使用

### 啟用 GPU 或 TPU

在 Colab 中啟用硬體加速：

1. 點擊選單列中的 **運行時（Runtime）** → **更改運行時類型（Change runtime type）**
2. 在 **硬體加速器（Hardware accelerator）** 下拉選單中選擇：
   - **None**：僅使用 CPU
   - **T4 GPU**：免費的 NVIDIA T4 GPU（約 15GB 顯存）
   - **TPU**：Google 的 Tensor Processing Unit

### 免費版硬體規格（2024）

| 資源 | 規格 | 限制 |
|------|------|------|
| GPU | NVIDIA T4 (15GB) | 連續使用限制 |
| RAM | 12-13 GB | 根據可用性 |
| 磁碟 | ~100 GB | 臨時存儲 |
| 連線時間 | 最多 12 小時 | 閒置 90 分鐘斷線 |

## 限制和最佳實踐

### 免費版限制

1. **使用時間限制**：
   - 最多連續運行 12 小時
   - 閒置 90 分鐘後自動斷線
   - 每日有 GPU 使用時數限制

2. **資源限制**：
   - GPU 記憶體有限（T4 約 15GB）

#### 4. 使用混合精度訓練

```python
# 使用自動混合精度（AMP）節省記憶體
from torch.cuda.amp import autocast, GradScaler

scaler = GradScaler()

for data, target in dataloader:
    optimizer.zero_grad()
    
    # 使用 autocast 自動選擇合適的精度
    with autocast():
        output = model(data)
        loss = criterion(output, target)
    
    # 使用 scaler 處理梯度
    scaler.scale(loss).backward()
    scaler.step(optimizer)
    scaler.update()
```

## Colab Pro 和 Pro+ 選項

### Colab Pro（約 $9.99/月）

Pro 訂閱提供以下升級：

- **更長的運行時間**：最多 24 小時連續運行
- **更好的 GPU**：優先訪問 V100 GPU
- **更多記憶體**：高達 32GB RAM
- **後台執行**：即使關閉瀏覽器也能繼續運行
- **更少限制**：更寬鬆的使用限制

### Colab Pro+（約 $49.99/月）

Pro+ 提供最高級別的訪問：

- **頂級 GPU**：A100 GPU（40GB 顯存）
- **更多資源**：高達 52GB RAM
- **更長運行時間**：可能超過 24 小時
- **更大儲存空間**：更多磁碟空間
- **優先級支持**：獲得更好的技術支持

### GPU 比較（2024）

| GPU 型號 | 顯存 | 計算能力 | 可用版本 | 適用場景 |
|---------|------|---------|----------|----------|
| T4 | 15GB | 8.5 TFLOPS (FP16) | 免費 | 小型模型訓練 |
| V100 | 16GB | 15.7 TFLOPS (FP16) | Pro | 中型模型訓練 |
| A100 | 40GB | 19.5 TFLOPS (FP16) | Pro+ | 大型模型訓練 |

### 選擇建議

- **學生/初學者**：免費版本足夠學習基礎知識
- **研究人員**：Pro 提供更穩定的環境和更好的 GPU
- **專業開發者**：Pro+ 適合需要訓練大型模型的情況
- **企業用戶**：考慮使用專門的雲端服務（AWS、GCP、Azure）
