# 选择服务器和GPU
:label:`sec_buy_gpu`

深度学习训练通常需要大量的计算。目前，GPU是深度学习最具成本效益的硬件加速器。与CPU相比，GPU更便宜，性能更高，通常超过一个数量级。此外，一台服务器可以支持多个GPU，高端服务器最多支持8个GPU。更典型的数字是工程工作站最多4个GPU，这是因为热量、冷却和电源需求会迅速增加，超出办公楼所能支持的范围。对于更大的部署，云计算（例如亚马逊的 P3、P4d、P5 和 G5 实例）是一个更实用的解决方案。

## 选择服务器

通常不需要购买具有多个线程的高端CPU，因为大部分计算都发生在GPU上。这就是说，由于Python中的全局解释器锁（GIL），CPU的单线程性能在有4-8个GPU的情况下可能很重要。所有的条件都是一样的，这意味着核数较少但时钟频率较高的CPU可能是更经济的选择。例如，当在6核4GHz和8核3.5GHz CPU之间进行选择时，前者更可取，即使其聚合速度较低。一个重要的考虑因素是，GPU使用大量的电能，从而释放大量的热量。这需要非常好的冷却和足够大的机箱来容纳GPU。如有可能，请遵循以下指南：

1. **电源**。GPU使用大量的电源。每个设备预计高达 350-600W（检查显卡的*峰值需求*而不是一般需求，因为高效代码可能会消耗大量能源）。現代高端 GPU 如 RTX 4090 可能需要高達 450W，專業級 GPU 如 H100 可能需要 700W。如果电源不能满足需求，系统会变得不稳定。
1. **机箱尺寸**。GPU很大，辅助电源连接器通常需要额外的空间。此外，大型机箱更容易冷却。現代 GPU 通常需要 2.5-3 插槽空間。
1. **GPU散热**。如果有大量的GPU，可能需要投资水冷。此外，即使风扇较少，也应以"公版设计"为目标，因为它们足够薄，可以在设备之间进气。当使用多风扇GPU，安装多个GPU时，它可能太厚而无法获得足够的空气。
1. **PCIe插槽**。在GPU之间来回移动数据（以及在GPU之间交换数据）需要大量带宽。建议使用 16 通道的 PCIe 4.0 或 5.0 插槽。当安装了多个GPU时，请务必仔细阅读主板说明，以确保在同时使用多个GPU时16$\times$带宽仍然可用，并且使用的是 PCIe 4.0/5.0，而不是用于附加插槽的 PCIe 3.0。在安装多个GPU的情况下，一些主板的带宽降级到8$\times$甚至4$\times$。这部分是由于CPU提供的PCIe通道数量限制。

### 系統記憶體建議（2024更新）

隨著深度學習模型和數據集的增長，對系統 RAM 的需求也在增加：

- **最低配置**：16GB RAM - 僅適合基本學習
- **推薦配置**：32GB RAM - 適合大多數深度學習任務
- **理想配置**：64GB+ RAM - 處理大型數據集和模型
- **專業配置**：128GB+ RAM - 大規模訓練和數據處理

**注意**：對於大型模型訓練，建議 RAM 容量至少是 GPU 總顯存的 2 倍。

简而言之，以下是构建深度学习服务器的一些建议（2024更新）。

## 構建深度學習服務器建議

### 入門配置（$1500-2500）
購買低功耗的入門級 GPU（適合深度學習的遊戲 GPU，功耗 150-200W）。

**建議配置**：
- **GPU**：RTX 4060 Ti 16GB
- **CPU**：Intel i5-13400 或 AMD Ryzen 5 7600
- **RAM**：32GB DDR4/DDR5
- **存儲**：500GB NVMe SSD + 2TB HDD
- **電源**：650W 80+ Gold
- **主板**：支持 PCIe 4.0

### 單 GPU 工作站（$2500-4000）
一個 4-6 核的 CPU 就足夠了。

**建議配置**：
- **GPU**：RTX 4080 16GB 或 RTX 4090 24GB
- **CPU**：Intel i7-13700K 或 AMD Ryzen 7 7800X3D
- **RAM**：64GB DDR5
- **存儲**：1TB NVMe SSD（主系統）+ 2TB NVMe SSD（數據）
- **電源**：850W 80+ Gold（RTX 4080）或 1000W 80+ Platinum（RTX 4090）
- **散熱**：高性能風冷或一體式水冷
- **機箱**：全塔式，良好的氣流

### 雙 GPU 配置（$4000-6000）
需要更好的散熱和電源管理。

**建議配置**：
- **GPU**：2× RTX 4080 16GB
- **CPU**：Intel i9-13900K 或 AMD Ryzen 9 7950X
- **RAM**：64-128GB DDR5
- **存儲**：2TB NVMe SSD（系統 + 快取）+ 4TB NVMe SSD（數據）
- **電源**：1200-1500W 80+ Platinum
- **主板**：支持 2× PCIe 4.0 x16 插槽（間距至少 2 個插槽）
- **散熱**：高端風冷或定制水冷
- **機箱**：超大全塔，多個風扇位

**重要注意事項**：
- 確保主板在兩個插槽同時使用時都能提供 x16 帶寬
- 兩個 GPU 之間需要足夠的空間以確保散熱
- 使用"公版"或渦輪式設計的 GPU 以改善多卡散熱

### 四 GPU 配置（$8000-15000）
專業級配置，需要特殊考慮。

**建議配置**：
- **GPU**：4× RTX 4090 24GB（公版設計）
- **CPU**：AMD Threadripper PRO 5975WX（32 核）或更高
  - 需要更多 PCIe 通道（64+ 通道）
  - 良好的單線程性能仍然重要
- **RAM**：128-256GB DDR4 ECC（推薦）
- **存儲**：
  - 2TB NVMe SSD（系統，PCIe 4.0）
  - 4TB NVMe SSD（工作快取，PCIe 4.0）
  - RAID 配置的 HDD 陣列用於長期存儲
- **電源**：2000W 80+ Titanium 或雙電源配置
- **主板**：
  - 支持 4× PCIe 4.0 x16 插槽
  - 可能需要 PLX 晶片進行 PCIe 通道多路復用
  - 專業工作站主板（如 ASUS Pro WS 系列）
- **散熱**：定制水冷系統（CPU + GPU）或專業機架散熱
- **機箱**：
  - 專業工作站機箱或 4U 機架式機箱
  - 優秀的氣流設計
  - 足夠的空間容納 4 個三槽 GPU

**特別注意**：
- 此配置會*非常吵且發熱*，不適合放在辦公桌下
- 可能需要專門的電路（20A 或更高）
- 考慮獨立的機房或數據中心託管
- 定期維護和監控至關重要

### 八 GPU 伺服器（$20000-50000+）
需要專業伺服器硬體。

**建議配置**：
- **GPU**：8× RTX 4090 或專業級 GPU（A100/H100）
- **機箱**：專用多 GPU 伺服器機箱
  - 4U 或 8U 機架式
  - 支持 GPU 的物理外形因素
  - 檢查電源線間隙和接線
- **電源**：
  - 雙冗餘電源（2+1 配置）
  - 每個 1600-2000W
  - 需要專業級電力供應
- **CPU**：雙路 Xeon 或 EPYC
  - 提供足夠的 PCIe 通道
  - 支持 ECC 記憶體
- **RAM**：256-512GB DDR4/DDR5 ECC
- **網路**：10GbE 或更快（建議 25GbE/100GbE）
- **冷卻**：
  - 專業機架冷卻系統
  - 可能需要液冷解決方案
  - 溫度監控系統

**重要警告**：
- 消費級 GPU（RTX 系列）與專業伺服器 GPU（A/H 系列）在物理外形、電源接口和冷卻設計上可能不兼容
- 在購買前務必驗證所有組件的兼容性
- 考慮使用雲端服務可能更經濟實惠

## 选择GPU

目前，AMD和NVIDIA是专用GPU的两大主要制造商。NVIDIA是第一个进入深度学习领域的公司，通过CUDA为深度学习框架提供更好的支持。因此，大多数买家选择NVIDIA GPU。

NVIDIA提供两种类型的GPU，针对个人用户（例如，通过 RTX 系列）和企业用户（通过其 A/H/L 系列）。这两种类型的GPU提供了相当的计算能力。但是，企业用户GPU通常使用强制（被动）冷却、更多内存和ECC（纠错）内存。这些GPU更适用于数据中心，通常成本是消费者GPU的 5-10 倍。

如果是一个拥有100个服务器的大公司，则应该考虑 NVIDIA A/H 系列，或者在云中使用GPU服务器。对于实验室或中小型公司（10-50 台伺服器），NVIDIA RTX 系列可能是最具成本效益的。也可以購買超微（Supermicro）或華碩（ASUS）等廠商的預配置伺服器，這些服務器可以有效地容納 4-8 個 GPU。

### GPU 性能參數

GPU供应商通常每一到两年发布一代，例如 2020 年發布的 RTX 3000（Ampere）系列和 2022 年發布的 RTX 4000（Ada Lovelace）系列。每个系列都提供几种不同的型号，提供不同的性能级别。GPU性能主要是以下參數的組合：

1. **計算能力**。通常大家會追求 32 位元浮點運算能力。16 位元浮點訓練（FP16）也進入主流，現代 GPU 支持 FP8、INT8 和 INT4 等低精度格式。Tensor Core 提供針對深度學習優化的混合精度運算，可將性能提升 2-8 倍。
   
2. **記憶體大小**。隨著模型變大或訓練期間使用的批次大小變大，將需要更多的 GPU 記憶體。檢查 HBM2/HBM2e/HBM3（高頻寬記憶體）與 GDDR6/GDDR6X 記憶體。HBM 速度更快但成本更高。
   
3. **記憶體頻寬**。當有足夠的記憶體頻寬時，才能最大限度地利用計算能力。現代高端 GPU 提供 500GB/s 到 3TB/s 的頻寬。

4. **互聯技術**。對於多 GPU 配置：
   - **NVLink**：NVIDIA 的高速 GPU 互聯（900GB/s - 1.8TB/s）
   - **PCIe 4.0/5.0**：標準接口，頻寬較低但廣泛支持

对于大多数用户，查看計算能力和記憶體大小就足夠了。請注意，許多 GPU 提供不同類型的加速。例如，NVIDIA 的 Tensor Core 將特定操作的速度提高了 2-16 倍。確保所使用的函式庫支持這一點。GPU 記憶體應不小於 8GB（12GB 更好，16GB 以上是理想的）。盡量避免將 GPU 也用於顯示 GUI（改用內建顯卡）。如果無法避免，請添加額外的 2GB RAM 以確保安全。

### GPU 記憶體需求指南

根據不同的應用場景，以下是 GPU 記憶體需求的參考：

| 應用場景 | 最低顯存 | 推薦顯存 | 理想顯存 |
|---------|---------|---------|---------|
| 學習基礎知識 | 6GB | 8-12GB | 16GB |
| 電腦視覺（ResNet、YOLO） | 8GB | 12-16GB | 24GB |
| NLP（BERT、GPT 微調） | 12GB | 16-24GB | 24GB+ |

## 雲端 vs. 本地部署考量

### 選擇本地 GPU 的情況：
- 長期頻繁使用（每月 >100 小時）
- 對數據隱私有嚴格要求
- 需要完全控制硬體配置
- 預算充足（一次性投資）

### 選擇雲端 GPU 的情況：
- 不定期使用或突發性需求
- 需要測試不同 GPU 配置
- 不想處理硬體維護
- 需要快速擴展（多節點訓練）
- 資本預算有限

### 成本分析示例（2024）

**本地 RTX 4090 工作站**：
- 初始成本：$3,500
- 分攤到 3 年：約 $100/月
- 適合每月使用 >100 小時

**雲端 GPU（AWS g5.xlarge，A10G）**：
- 按需價格：約 $1.00/小時
- 每月 100 小時：$100/月
- 適合不定期使用

**結論**：如果每月使用超過 100-150 小時，本地 GPU 更經濟實惠。

## AMD GPU 考量

雖然 NVIDIA 主導深度學習市場，AMD 正在快速發展：

### AMD Radeon RX 7000 系列（2024）
- **RX 7900 XTX**：24GB GDDR6，性能接近 RTX 4080
- **優勢**：價格較低，大顯存
- **劣勢**：ROCm 生態系統不如 CUDA 成熟

### 考慮 AMD 的情況：
- 預算有限但需要大顯存
- 使用支持 ROCm 的框架（PyTorch、TensorFlow）
- 開源軟體棧優先

### 建議：
對於初學者和大多數應用，NVIDIA GPU 仍是最安全的選擇，因為生態系統支持更完善。

## 小结

* 在構建服務器時，請注意電源、PCIe 總線通道、CPU 單線程速度和散熱
* 2024 年推薦的消費級 GPU：RTX 4060 Ti（入門）、RTX 4070 Ti（中階）、RTX 4080（高階）、RTX 4090（頂級）
* 系統記憶體建議：最低 16GB，推薦 32GB，理想 64GB 或更多
* GPU 架構：Ada Lovelace（消費級）和 Hopper（專業級）是當前主流
* 如果可能，應該購買最新一代的 GPU，以獲得更好的能效和新功能
* 對於大規模部署，使用雲端或專業數據中心更實用
* 多 GPU 配置需要特別注意散熱、電源和 PCIe 頻寬
* 高密度服務器可能不與所有 GPU 兼容，購買前請檢查機械和散熱規格
* 為提高效率，請使用 FP16、FP8 或更低精度的訓練
* 考慮總擁有成本（TCO），包括電費、維護和折舊
