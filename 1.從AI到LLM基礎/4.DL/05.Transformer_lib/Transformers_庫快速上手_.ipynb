2. 使用 pipeline 進行情感分析
以下範例展示如何使用 pipeline 快速執行情感分析任務。

````python
from transformers import pipeline

# 建立情感分析的 pipeline（會自動下載預訓練模型與分詞器）
classifier = pipeline("sentiment-analysis")

# 單筆輸入範例
result = classifier("We are very happy to show you the Transformers library.")
print(result)

# 多筆輸入範例
results = classifier([
    "We are very happy to show you the Transformers library.",
    "We hope you don't hate it."
])
for res in results:
    print(f"label: {res['label']}, score: {round(res['score'], 4)}")
````

3. 使用 Speech Recognition Pipeline 搭配資料集
此範例示範如何利用語音識別 pipeline 搭配 Hugging Face Datasets 讀取音頻資料並執行推理。

````python
import torch
from transformers import pipeline
from datasets import load_dataset, Audio

# 建立語音識別 pipeline，使用 facebook/wav2vec2-base-960h 模型
speech_recognizer = pipeline("automatic-speech-recognition", model="facebook/wav2vec2-base-960h")

# 載入 PolyAI/minds14 音頻資料集 (此處選用 "en-US" 分割)
dataset = load_dataset("PolyAI/minds14", name="en-US", split="train")

# 將資料集中的 audio 欄位轉型以符合模型所需的採樣率
dataset = dataset.cast_column("audio", Audio(sampling_rate=speech_recognizer.feature_extractor.sampling_rate))

# 對前 4 筆音頻資料執行推理
result = speech_recognizer(dataset[:4]["audio"])
print([d["text"] for d in result])
````

````python
from transformers import pipeline

# 建立零樣本分類 pipeline
zero_shot_classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

# 待分類文本
text = "I love using Transformers library. It makes NLP so easy!"

# 候選標籤
candidate_labels = ["technology", "sports", "politics", "entertainment", "education"]

# 執行分類
result = zero_shot_classifier(text, candidate_labels)

print(f"文本: {text}\n")
print("分類結果：")
for label, score in zip(result['labels'], result['scores']):
    print(f"  {label}: {score:.4f}")

print("\n" + "="*50 + "\n")

# 多標籤分類示例
text2 = "This new programming course teaches Python and machine learning concepts."
candidate_labels2 = ["education", "technology", "business", "health"]

result2 = zero_shot_classifier(text2, candidate_labels2, multi_label=True)

print(f"文本: {text2}\n")
print("多標籤分類結果：")
for label, score in zip(result2['labels'], result2['scores']):
    print(f"  {label}: {score:.4f}")
````

## 2.6 零樣本分類 (Zero-shot Classification)

零樣本分類可以在沒有訓練數據的情況下對文本進行分類。

````python
from transformers import pipeline

# 英文翻譯成德文
en_to_de = pipeline("translation_en_to_de", model="Helsinki-NLP/opus-mt-en-de")

text_en = "Hugging Face is creating a tool that democratizes AI."
result = en_to_de(text_en)
print(f"英文: {text_en}")
print(f"德文: {result[0]['translation_text']}")

print("\n" + "="*50 + "\n")

# 英文翻譯成法文
en_to_fr = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")

texts = [
    "Machine learning is a subset of artificial intelligence.",
    "Deep learning models require large amounts of data."
]

print("批次翻譯（英文 -> 法文）：")
for text in texts:
    result = en_to_fr(text)
    print(f"  EN: {text}")
    print(f"  FR: {result[0]['translation_text']}\n")
````

## 2.5 機器翻譯 (Translation)

使用預訓練的翻譯模型進行語言翻譯。

````python
from transformers import pipeline

# 建立摘要 pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# 長文本範例
article = """
Artificial intelligence (AI) is intelligence demonstrated by machines, in contrast to the natural 
intelligence displayed by humans and animals. Leading AI textbooks define the field as the study 
of "intelligent agents": any device that perceives its environment and takes actions that maximize 
its chance of successfully achieving its goals. Colloquially, the term "artificial intelligence" 
is often used to describe machines (or computers) that mimic "cognitive" functions that humans 
associate with the human mind, such as "learning" and "problem solving".

As machines become increasingly capable, tasks considered to require "intelligence" are often 
removed from the definition of AI, a phenomenon known as the AI effect. A quip in Tesler's Theorem 
says "AI is whatever hasn't been done yet." For instance, optical character recognition is frequently 
excluded from things considered to be AI, having become a routine technology.

Modern machine learning approaches, which include deep learning, are particularly effective for 
natural language processing, computer vision, speech recognition, and many other applications.
"""

# 生成摘要
summary = summarizer(article, max_length=100, min_length=30, do_sample=False)

print("原文長度:", len(article.split()))
print("\n摘要:")
print(summary[0]['summary_text'])
print("\n摘要長度:", len(summary[0]['summary_text'].split()))
````

## 2.4 文本摘要 (Summarization)

文本摘要可以將長文本濃縮成簡短的摘要。

````python
from transformers import pipeline

# 建立問答 pipeline
qa_pipeline = pipeline("question-answering")

# 準備上下文和問題
context = """
The Transformer architecture was introduced in the paper "Attention is All You Need" by Vaswani et al. in 2017.
It revolutionized natural language processing by relying entirely on attention mechanisms, 
dispensing with recurrence and convolutions entirely. The model achieves state-of-the-art results 
on machine translation and other NLP tasks.
"""

# 提出多個問題
questions = [
    "Who introduced the Transformer architecture?",
    "When was the Transformer introduced?",
    "What makes the Transformer special?"
]

print("問答系統示範：\n")
for question in questions:
    result = qa_pipeline(question=question, context=context)
    print(f"Q: {question}")
    print(f"A: {result['answer']} (信心度: {result['score']:.4f})")
    print(f"   位置: {result['start']}-{result['end']}\n")
````

## 2.3 問答系統 (Question Answering)

問答系統可以從給定的上下文中回答問題。

````python
from transformers import pipeline

# 建立 NER pipeline
ner_pipeline = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")

# 示例文本
text = "Apple Inc. was founded by Steve Jobs in Cupertino, California."

# 執行 NER
entities = ner_pipeline(text)

print("識別出的實體：")
for entity in entities:
    print(f"  - {entity['word']}: {entity['entity_group']} (信心度: {entity['score']:.4f})")

# 中文 NER 範例（使用中文模型）
# 註：以下使用英文模型示範，實際應用可選擇中文 NER 模型
print("\n另一個範例：")
text2 = "Microsoft was founded by Bill Gates and Paul Allen in Seattle."
entities2 = ner_pipeline(text2)
for entity in entities2:
    print(f"  - {entity['word']}: {entity['entity_group']} (信心度: {entity['score']:.4f})")
````

````python
import torch
from transformers import pipeline, AutoTokenizer, AutoModelForSequenceClassification

# 1. 檢查 GPU 可用性
print("=== 1. GPU 配置檢查 ===\n")

print(f"CUDA 是否可用: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA 版本: {torch.version.cuda}")
    print(f"GPU 數量: {torch.cuda.device_count()}")
    print(f"當前 GPU: {torch.cuda.current_device()}")
    print(f"GPU 名稱: {torch.cuda.get_device_name(0)}")
    print(f"GPU 記憶體: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.2f} GB")
else:
    print("未檢測到 GPU，將使用 CPU")

print("\n" + "="*50 + "\n")

# 2. 指定設備運行 Pipeline
print("=== 2. 指定設備運行 ===\n")

# 方法 1: 使用 device 參數
device = 0 if torch.cuda.is_available() else -1  # 0 表示 GPU 0，-1 表示 CPU
classifier = pipeline("sentiment-analysis", device=device)

print(f"✓ Pipeline 運行在: {'GPU' if device >= 0 else 'CPU'}")

# 測試推理
result = classifier("This is amazing!")
print(f"測試結果: {result}")

print("\n" + "="*50 + "\n")

# 3. 混合精度推理 (FP16)
print("=== 3. 混合精度推理 (加速 2-3 倍) ===\n")

if torch.cuda.is_available():
    import time
    
    # FP32 推理
    model_name = "distilbert-base-uncased-finetuned-sst-2-english"
    model_fp32 = AutoModelForSequenceClassification.from_pretrained(model_name).to("cuda")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    
    text = "This is a test sentence for performance comparison."
    inputs = tokenizer(text, return_tensors="pt").to("cuda")
    
    # 預熱
    for _ in range(10):
        with torch.no_grad():
            _ = model_fp32(**inputs)
    
    # FP32 推理測試
    start = time.time()
    for _ in range(100):
        with torch.no_grad():
            outputs = model_fp32(**inputs)
    time_fp32 = time.time() - start
    
    # FP16 推理
    model_fp16 = AutoModelForSequenceClassification.from_pretrained(model_name).to("cuda").half()
    inputs_fp16 = {k: v.to("cuda") for k, v in inputs.items()}
    
    # 預熱
    for _ in range(10):
        with torch.no_grad():
            _ = model_fp16(**inputs_fp16)
    
    # FP16 推理測試
    start = time.time()
    for _ in range(100):
        with torch.no_grad():
            outputs = model_fp16(**inputs_fp16)
    time_fp16 = time.time() - start
    
    print(f"FP32 推理時間: {time_fp32:.4f} 秒")
    print(f"FP16 推理時間: {time_fp16:.4f} 秒")
    print(f"加速比: {time_fp32 / time_fp16:.2f}x")
    
    # 記憶體使用比較
    print(f"\nFP32 模型大小: ~{sum(p.numel() * 4 for p in model_fp32.parameters()) / 1024**2:.2f} MB")
    print(f"FP16 模型大小: ~{sum(p.numel() * 2 for p in model_fp16.parameters()) / 1024**2:.2f} MB")
else:
    print("需要 GPU 才能進行混合精度推理")

print("\n" + "="*50 + "\n")

# 4. 模型量化 (8-bit)
print("=== 4. 模型量化 (節省記憶體) ===\n")

try:
    # 需要安裝 bitsandbytes: pip install bitsandbytes
    from transformers import AutoModelForCausalLM
    
    print("載入 8-bit 量化模型（需要 GPU 和 bitsandbytes 庫）...")
    # 注意：這需要較大的模型才能看到明顯效果
    # model_8bit = AutoModelForCausalLM.from_pretrained(
    #     "gpt2",
    #     load_in_8bit=True,
    #     device_map="auto"
    # )
    # print("✓ 8-bit 模型已載入")
    print("（示例代碼已註釋，取消註釋以執行）")
except ImportError:
    print("✗ 需要安裝 bitsandbytes 庫: pip install bitsandbytes")

print("\n" + "="*50 + "\n")

# 5. 批次大小優化
print("=== 5. 動態批次大小優化 ===\n")

def find_optimal_batch_size(model, tokenizer, texts, max_batch_size=32):
    """自動尋找最佳批次大小"""
    device = "cuda" if torch.cuda.is_available() else "cpu"
    model = model.to(device)
    
    batch_size = max_batch_size
    while batch_size > 0:
        try:
            # 測試批次處理
            batch_texts = texts[:batch_size]
            inputs = tokenizer(batch_texts, return_tensors="pt", padding=True, truncation=True)
            inputs = {k: v.to(device) for k, v in inputs.items()}
            
            with torch.no_grad():
                _ = model(**inputs)
            
            print(f"✓ 最佳批次大小: {batch_size}")
            return batch_size
        except RuntimeError as e:
            if "out of memory" in str(e):
                batch_size = batch_size // 2
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                print(f"  嘗試批次大小: {batch_size}")
            else:
                raise e
    
    return 1

# 測試
test_texts = ["This is a test sentence."] * 100
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")

if torch.cuda.is_available():
    optimal_bs = find_optimal_batch_size(model, tokenizer, test_texts)
else:
    print("在 CPU 上運行，使用預設批次大小 8")

print("\n" + "="*50 + "\n")

# 6. 推理時的記憶體優化技巧
print("=== 6. 記憶體優化技巧總結 ===\n")

optimization_tips = """
✓ 使用 torch.no_grad() 禁用梯度計算
✓ 使用混合精度 (FP16/BF16)
✓ 使用模型量化 (8-bit/4-bit)
✓ 批次處理而非逐一處理
✓ 使用 gradient_checkpointing 訓練大模型
✓ 定期清理 GPU 快取 (torch.cuda.empty_cache())
✓ 使用 device_map="auto" 自動分配設備
✓ 考慮使用 ONNX Runtime 進行部署優化
"""

print(optimization_tips)

print("✅ GPU 配置與優化示範完成！")
````

## 13. GPU 配置與優化

展示如何配置 GPU 並優化模型性能。

````python
from transformers import pipeline, AutoTokenizer, AutoModelForSequenceClassification
import torch

# 1. 處理模型載入錯誤
print("=== 1. 安全載入模型 ===\n")

def safe_load_pipeline(task, model_name=None, max_retries=3):
    """安全載入 pipeline，包含錯誤處理和重試機制"""
    for attempt in range(max_retries):
        try:
            if model_name:
                pipe = pipeline(task, model=model_name)
            else:
                pipe = pipeline(task)
            print(f"✓ 成功載入 {task} pipeline")
            return pipe
        except Exception as e:
            print(f"✗ 嘗試 {attempt + 1} 失敗: {str(e)}")
            if attempt == max_retries - 1:
                print(f"✗ 載入失敗，已達最大重試次數")
                return None
    return None

# 示例使用
classifier = safe_load_pipeline("sentiment-analysis")

print("\n" + "="*50 + "\n")

# 2. 處理輸入長度限制
print("=== 2. 處理長文本輸入 ===\n")

model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

long_text = " ".join(["This is a very long text."] * 200)  # 創建超長文本

def safe_inference(text, tokenizer, model, max_length=512):
    """安全執行推理，自動處理長度限制"""
    try:
        # 方法 1: 截斷
        inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=max_length)
        
        # 檢查設備
        device = "cuda" if torch.cuda.is_available() else "cpu"
        model = model.to(device)
        inputs = {k: v.to(device) for k, v in inputs.items()}
        
        # 推理
        with torch.no_grad():
            outputs = model(**inputs)
        
        # 取得預測
        predictions = torch.nn.functional.softmax(outputs.logits, dim=-1)
        predicted_class = torch.argmax(predictions, dim=-1).item()
        confidence = predictions[0][predicted_class].item()
        
        return {
            "label": model.config.id2label[predicted_class],
            "confidence": confidence,
            "truncated": len(tokenizer.tokenize(text)) > max_length
        }
    except Exception as e:
        print(f"✗ 推理錯誤: {str(e)}")
        return None

result = safe_inference(long_text, tokenizer, model)
if result:
    print(f"✓ 預測標籤: {result['label']}")
    print(f"✓ 信心度: {result['confidence']:.4f}")
    print(f"✓ 文本被截斷: {result['truncated']}")

print("\n" + "="*50 + "\n")

# 3. 批次處理優化
print("=== 3. 高效批次推理 ===\n")

texts = [
    "I love this product!",
    "This is terrible.",
    "Not bad, could be better.",
    "Amazing experience!",
    "Disappointing quality."
]

# 錯誤方式：逐一處理（慢）
print("逐一處理（慢）:")
import time
start = time.time()
results_slow = []
for text in texts:
    result = classifier(text)
    results_slow.append(result)
time_slow = time.time() - start
print(f"  耗時: {time_slow:.4f} 秒")

# 正確方式：批次處理（快）
print("\n批次處理（快）:")
start = time.time()
results_fast = classifier(texts, batch_size=8)
time_fast = time.time() - start
print(f"  耗時: {time_fast:.4f} 秒")
print(f"  加速比: {time_slow / time_fast:.2f}x")

print("\n" + "="*50 + "\n")

# 4. GPU/CPU 記憶體管理
print("=== 4. 記憶體管理最佳實踐 ===\n")

def clear_memory():
    """清理 GPU/CPU 記憶體"""
    import gc
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        print("✓ GPU 記憶體已清理")
    else:
        print("✓ CPU 記憶體已清理")

# 示例：載入大模型前清理記憶體
clear_memory()

# 檢查 GPU 記憶體使用
if torch.cuda.is_available():
    print(f"\nGPU 記憶體使用:")
    print(f"  已分配: {torch.cuda.memory_allocated() / 1024**2:.2f} MB")
    print(f"  已保留: {torch.cuda.memory_reserved() / 1024**2:.2f} MB")

print("\n" + "="*50 + "\n")

# 5. 設定隨機種子以確保可重現性
print("=== 5. 確保結果可重現 ===\n")

def set_seed(seed=42):
    """設定所有隨機種子"""
    import random
    import numpy as np
    
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    
    print(f"✓ 隨機種子已設定為 {seed}")

set_seed(42)

print("\n" + "="*50 + "\n")

# 6. 離線模式（預先下載模型）
print("=== 6. 離線使用模型 ===\n")

print("步驟 1: 預先下載並保存模型")
save_path = "./my_saved_model"

# 下載並保存
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")

model.save_pretrained(save_path)
tokenizer.save_pretrained(save_path)
print(f"✓ 模型已保存至 {save_path}")

print("\n步驟 2: 從本地載入（離線模式）")
# 離線載入
model_offline = AutoModelForSequenceClassification.from_pretrained(save_path, local_files_only=True)
tokenizer_offline = AutoTokenizer.from_pretrained(save_path, local_files_only=True)
print(f"✓ 已從本地載入模型")

print("\n✅ 所有最佳實踐示範完成！")
````

9. 使用 Trainer 進行模型訓練（PyTorch 範例）
以下範例展示如何使用 Hugging Face 提供的 Trainer 來訓練模型。此處以 Rotten Tomatoes 資料集作為示例。

````python
from transformers import TrainingArguments, Trainer, DataCollatorWithPadding
from datasets import load_dataset

# 載入 Rotten Tomatoes 資料集
dataset = load_dataset("rotten_tomatoes")

# 定義分詞函式
def tokenize_dataset(example):
    return tokenizer(example["text"], padding="max_length", truncation=True)

# 對資料集進行分詞處理
dataset = dataset.map(tokenize_dataset, batched=True)

# 建立 DataCollator (將資料整理成統一長度的批次)
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# 設定訓練參數
training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=2,
)

# 建立 Trainer 實例
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    data_collator=data_collator,
)

# 執行訓練（如欲測試，可取消下行註解）
trainer.train()
````
