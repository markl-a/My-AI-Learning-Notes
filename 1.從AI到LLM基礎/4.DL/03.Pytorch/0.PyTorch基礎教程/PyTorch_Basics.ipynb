## 1. Tensor 基礎操作

Tensor 是 PyTorch 中最基本的資料結構,類似於 NumPy 的 ndarray,但可以在 GPU 上運行。

### 1.1 創建 Tensor

```python
# 從列表創建
t1 = torch.tensor([1, 2, 3])
print("從列表創建:", t1)

# 從 NumPy 數組創建
arr = np.array([[1, 2], [3, 4]])
t2 = torch.from_numpy(arr)
print("從 NumPy 創建:\n", t2)

# 創建全零張量
t3 = torch.zeros(2, 3)
print("全零張量:\n", t3)

# 創建全一張量
t4 = torch.ones(2, 3)
print("全一張量:\n", t4)

# 創建隨機張量
t5 = torch.rand(2, 3)  # 均勻分佈 [0, 1)
print("隨機張量 (均勻分佈):\n", t5)

t6 = torch.randn(2, 3)  # 標準正態分佈
print("隨機張量 (正態分佈):\n", t6)

# 創建指定範圍的張量
t7 = torch.arange(0, 10, 2)
print("指定範圍:", t7)

t8 = torch.linspace(0, 1, 5)
print("線性分佈:", t8)
```

### 1.2 Tensor 屬性

```python
x = torch.randn(3, 4)

print("張量:\n", x)
print("形狀:", x.shape)  # 或 x.size()
print("維度:", x.ndim)
print("數據類型:", x.dtype)
print("設備:", x.device)
print("元素總數:", x.numel())
```

### 1.3 Tensor 運算

```python
# 創建示例張量
a = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
b = torch.tensor([[5.0, 6.0], [7.0, 8.0]])

# 加法
print("加法:\n", a + b)
print("加法 (方法):\n", torch.add(a, b))

# 減法
print("減法:\n", a - b)

# 乘法 (逐元素)
print("逐元素乘法:\n", a * b)

# 矩陣乘法
print("矩陣乘法:\n", a @ b)
print("矩陣乘法 (方法):\n", torch.matmul(a, b))

# 除法
print("除法:\n", a / b)

# 幂運算
print("平方:\n", a ** 2)

# 其他運算
print("轉置:\n", a.T)
print("求和:", a.sum())
print("按行求和:", a.sum(dim=0))
print("按列求和:", a.sum(dim=1))
print("平均值:", a.mean())
print("最大值:", a.max())
print("最小值:", a.min())
```

### 1.4 Tensor 形狀操作

```python
x = torch.randn(2, 3, 4)
print("原始形狀:", x.shape)

# view: 改變形狀 (需要連續內存)
x_view = x.view(2, 12)
print("view 後:", x_view.shape)

# reshape: 改變形狀 (更靈活)
x_reshape = x.reshape(6, 4)
print("reshape 後:", x_reshape.shape)

# 自動推斷維度
x_auto = x.view(2, -1)
print("自動推斷:", x_auto.shape)

# squeeze: 移除大小為1的維度
x_squeeze = torch.randn(1, 3, 1, 4)
print("squeeze 前:", x_squeeze.shape)
print("squeeze 後:", x_squeeze.squeeze().shape)

# unsqueeze: 添加維度
x_unsqueeze = torch.randn(3, 4)
print("unsqueeze 前:", x_unsqueeze.shape)
print("unsqueeze 後:", x_unsqueeze.unsqueeze(0).shape)

# transpose: 交換維度
x_transpose = torch.randn(2, 3, 4)
print("transpose 前:", x_transpose.shape)
print("transpose 後:", x_transpose.transpose(0, 1).shape)

# permute: 重排維度
print("permute 後:", x_transpose.permute(2, 0, 1).shape)
```

### 1.5 索引和切片

```python
x = torch.arange(12).reshape(3, 4)
print("原始張量:\n", x)

# 基本索引
print("第一行:", x[0])
print("第一列:", x[:, 0])
print("第一行第二列:", x[0, 1])

# 切片
print("前兩行:\n", x[:2])
print("前兩列:\n", x[:, :2])
print("子矩陣:\n", x[1:, 1:3])

# 布爾索引
mask = x > 5
print("大於5的元素:", x[mask])

# 高級索引
indices = torch.tensor([0, 2])
print("選擇第0和第2行:\n", x[indices])
```

### 1.6 GPU 加速

```python
# 檢查 CUDA 是否可用
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"使用設備: {device}")

# 創建 CPU 張量
x_cpu = torch.randn(3, 3)
print("CPU 張量:", x_cpu.device)

# 移動到 GPU
if torch.cuda.is_available():
    x_gpu = x_cpu.to(device)
    # 或者: x_gpu = x_cpu.cuda()
    print("GPU 張量:", x_gpu.device)
    
    # GPU 上的運算
    y_gpu = x_gpu * 2
    
    # 移回 CPU
    y_cpu = y_gpu.cpu()
    print("結果:\n", y_cpu)
else:
    print("GPU 不可用,使用 CPU")
```

## 2. 自動微分 (Autograd)

PyTorch 的自動微分系統是其核心功能之一,用於自動計算梯度。

### 2.1 基本概念

```python
# requires_grad=True 表示需要計算梯度
x = torch.tensor([2.0], requires_grad=True)
print("x:", x)
print("requires_grad:", x.requires_grad)

# 前向傳播
y = x ** 2 + 2 * x + 1
print("y:", y)

# 反向傳播計算梯度
y.backward()

# 查看梯度 dy/dx = 2x + 2
print("x 的梯度:", x.grad)
print("理論值:", 2 * x.item() + 2)
```

### 2.2 多變量梯度

```python
# 創建需要梯度的張量
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = torch.tensor([4.0, 5.0, 6.0], requires_grad=True)

# 前向傳播
z = (x * y).sum()
print("z:", z)

# 反向傳播
z.backward()

print("x 的梯度:", x.grad)  # dz/dx = y
print("y 的梯度:", y.grad)  # dz/dy = x
```

### 2.3 梯度累積

```python
x = torch.tensor([2.0], requires_grad=True)

# 第一次計算
y1 = x ** 2
y1.backward()
print("第一次梯度:", x.grad)

# 第二次計算 (梯度會累積)
y2 = x ** 3
y2.backward()
print("累積後的梯度:", x.grad)

# 清零梯度
x.grad.zero_()
print("清零後的梯度:", x.grad)

# 重新計算
y3 = x ** 2
y3.backward()
print("新的梯度:", x.grad)
```

### 2.4 停止梯度計算

```python
# 方法 1: torch.no_grad()
x = torch.tensor([2.0], requires_grad=True)

with torch.no_grad():
    y = x ** 2
    print("y requires_grad:", y.requires_grad)

# 方法 2: detach()
x = torch.tensor([2.0], requires_grad=True)
y = x ** 2
y_detached = y.detach()
print("y_detached requires_grad:", y_detached.requires_grad)

# 方法 3: requires_grad_(False)
x = torch.tensor([2.0], requires_grad=True)
x.requires_grad_(False)
print("x requires_grad:", x.requires_grad)
```

### 2.5 實際應用:線性回歸

```python
# 生成數據: y = 2x + 1 + noise
torch.manual_seed(42)
X = torch.randn(100, 1)
y_true = 2 * X + 1 + 0.1 * torch.randn(100, 1)

# 初始化參數
w = torch.randn(1, 1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

# 訓練參數
learning_rate = 0.01
n_epochs = 100

for epoch in range(n_epochs):
    # 前向傳播
    y_pred = X @ w + b
    
    # 計算損失
    loss = ((y_pred - y_true) ** 2).mean()
    
    # 反向傳播
    loss.backward()
    
    # 更新參數
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad
        
        # 清零梯度
        w.grad.zero_()
        b.grad.zero_()
    
    if (epoch + 1) % 20 == 0:
        print(f"Epoch [{epoch+1}/{n_epochs}], Loss: {loss.item():.4f}")

print(f"\n學習到的參數: w={w.item():.4f}, b={b.item():.4f}")
print(f"真實參數: w=2.0000, b=1.0000")
```

## 3. 神經網路基礎

使用 `nn.Module` 構建神經網路。

### 3.1 簡單的線性模型

```python
class LinearRegression(nn.Module):
    def __init__(self, input_size, output_size):
        super(LinearRegression, self).__init__()
        self.linear = nn.Linear(input_size, output_size)
    
    def forward(self, x):
        return self.linear(x)

# 創建模型
model = LinearRegression(1, 1)
print("模型結構:\n", model)

# 查看參數
for name, param in model.named_parameters():
    print(f"{name}: {param.shape}")
```

### 3.2 多層感知機 (MLP)

```python
class MLP(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_size, output_size)
    
    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

# 創建模型
model = MLP(10, 20, 5)
print("模型結構:\n", model)

# 測試前向傳播
x = torch.randn(32, 10)  # batch_size=32, input_size=10
output = model(x)
print(f"\n輸入形狀: {x.shape}")
print(f"輸出形狀: {output.shape}")
```

### 3.3 使用 Sequential

```python
# 方法 1: 順序定義
model = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Linear(20, 5)
)

print("方法 1:\n", model)

# 方法 2: 使用 OrderedDict
from collections import OrderedDict

model = nn.Sequential(OrderedDict([
    ('fc1', nn.Linear(10, 20)),
    ('relu1', nn.ReLU()),
    ('fc2', nn.Linear(20, 5))
]))

print("\n方法 2:\n", model)
```

## 4. 資料處理

使用 `Dataset` 和 `DataLoader` 處理資料。

### 4.1 自定義 Dataset

```python
class CustomDataset(Dataset):
    def __init__(self, data, labels):
        self.data = data
        self.labels = labels
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]

# 創建數據
data = torch.randn(100, 10)
labels = torch.randint(0, 2, (100,))

# 創建 Dataset
dataset = CustomDataset(data, labels)
print(f"Dataset 大小: {len(dataset)}")
print(f"第一個樣本: {dataset[0]}")
```

### 4.2 使用 DataLoader

```python
# 創建 DataLoader
dataloader = DataLoader(
    dataset,
    batch_size=16,
    shuffle=True,
    num_workers=0  # Windows 用戶可能需要設為 0
)

# 迭代數據
for batch_idx, (batch_data, batch_labels) in enumerate(dataloader):
    print(f"Batch {batch_idx}: data shape={batch_data.shape}, labels shape={batch_labels.shape}")
    if batch_idx >= 2:  # 只打印前3個batch
        break
```

## 5. 訓練循環

完整的訓練流程示例。

```python
# 1. 準備數據
torch.manual_seed(42)
X_train = torch.randn(1000, 10)
y_train = (X_train.sum(dim=1) > 0).long()

train_dataset = CustomDataset(X_train, y_train)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)

# 2. 定義模型
model = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Linear(20, 2)
)

# 3. 定義損失函數和優化器
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# 4. 訓練循環
n_epochs = 10
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

for epoch in range(n_epochs):
    model.train()  # 設置為訓練模式
    total_loss = 0
    correct = 0
    total = 0
    
    for batch_data, batch_labels in train_loader:
        # 移動到設備
        batch_data = batch_data.to(device)
        batch_labels = batch_labels.to(device)
        
        # 前向傳播
        outputs = model(batch_data)
        loss = criterion(outputs, batch_labels)
        
        # 反向傳播和優化
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        # 統計
        total_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        total += batch_labels.size(0)
        correct += (predicted == batch_labels).sum().item()
    
    # 打印訓練信息
    avg_loss = total_loss / len(train_loader)
    accuracy = 100 * correct / total
    print(f"Epoch [{epoch+1}/{n_epochs}], Loss: {avg_loss:.4f}, Accuracy: {accuracy:.2f}%")

print("\n訓練完成!")
```
